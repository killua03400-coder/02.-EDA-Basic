# 따릉이 대여 데이터 — 기초 EDA 리포트

## 1. 데이터 개요
- 행/열: (181행, 5열) -> 중복행 1개
- 기간: 2025-01-01 ~ 2025-06-29
- 주요 컬럼: date, rentals, avg_temp, rainfall, station_area

## 2. 구조 진단 (shape / info / describe)
- 자료형 요약: (날짜 열은 datetime으로 인식됨 / rentals·avg_temp는 float, rainfall은 int)
- 수치 요약에서 눈에 띈 점: (rentals 평균 4,532, avg_temp 최댓값 250.0 → 사람이 사는 곳에서
  나올 수 없는 기온이라 이상치 의심)

## 3. 품질 진단 (결측 / 중복 / 표기)
- 결측: rentals 8건  -> (181−8) 총 173건
- 중복: 1건
- 표기 혼재: station_area에 ' 강남 '(공백) 존재 → '강남'과 따로 집계될 위험

## 4. 패턴 (시각화에서 읽은 것)
- 추이: (계절성 — 봄~여름 증가 등) 1월부터 대여 수가 늘다가 중반 이후 다시 줄어듭니다.
- 관계: (기온과 대여 수의 양의 관계 등) 기온이 15도쯤까지는 대여가 늘다가 그 이상에서는 오히려 줄어드는 모양이었습니다. 완전한 우상향이 아니라 산 모양에 가깝게 보였습니다.

## 5. 다음 분석 제안
- (이상치/결측을 어떻게 처리할지, 추가로 보고 싶은 것) 
- avg_temp 250은 측정 오류로 보고 제외하거나 결측으로 바꾼다. 나머지 값이 최대 31.8이라 이 한 값만      동떨어져 있기 때문.
- rentals 결측 8건은 전체의 약 4%라 우선 제외하고, 제외 전후 평균 차이를 확인한다.
- 중복 1건은 같은 행이 두 번 들어간 것이므로 하나만 남긴다.
- station_area는 공백을 제거해 '강남'으로 통일한다.
